```python
import torch
import torchvision.models as models
import torchvision.transforms as transforms
from torchvision.datasets import ImageFolder
from torch.utils.data import DataLoader
import os
import pandas as pd
from tqdm import tqdm
from PIL import Image
import numpy as np
```

```python
DATASET_PATH = r"C:\Users\moham\OneDrive\Desktop\KFUPM\Courses\242\AML\Project\ImageDataset"
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
BATCH_SIZE = 1  # Use 1 to keep track of individual filenames
print(DEVICE)
```

```python
def get_feature_extractor(model_name):
    if model_name == "mobilenetv2":
        model = models.mobilenet_v2(pretrained=True).features
        feature_dim = 1280

    elif model_name == "resnet18":
        base = models.resnet18(pretrained=True)
        model = torch.nn.Sequential(*list(base.children())[:-1])  # remove FC
        feature_dim = 512

    elif model_name == "efficientnet_b0":
        base = models.efficientnet_b0(pretrained=True)
        model = torch.nn.Sequential(*list(base.children())[:-1])  # remove classifier
        feature_dim = 1280

    elif model_name == "densenet121":
        base = models.densenet121(pretrained=True)
        model = torch.nn.Sequential(base.features, torch.nn.AdaptiveAvgPool2d((1, 1)))
        feature_dim = 1024

    elif model_name == "vgg16":
        base = models.vgg16(pretrained=True)
        model = base.features  # remove classifier
        model.add_module("global_avg_pool", torch.nn.AdaptiveAvgPool2d((1, 1)))
        feature_dim = 512

    else:
        raise ValueError(f"Model {model_name} not supported.")

    model.eval()
    return model.to(DEVICE), feature_dim
```

```python
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],  # RGB pretrained model stats
                         std=[0.229, 0.224, 0.225])
])

# Custom dataset to keep filenames
class ImageDatasetWithPath(ImageFolder):
    def __getitem__(self, index):
        original_tuple = super().__getitem__(index)
        path, _ = self.samples[index]
        filename = os.path.basename(path)
        return original_tuple + (filename,)  # (image_tensor, class_label, filename)
```

```python
dataset = ImageDatasetWithPath(DATASET_PATH, transform=transform)
dataloader = DataLoader(dataset, batch_size=BATCH_SIZE, shuffle=False)
```

```python
model_names = ["mobilenetv2", "resnet18", "efficientnet_b0", "densenet121", "vgg16"]
feature_dfs = []

# Disable gradient tracking
with torch.no_grad():
    for model_name in model_names:
        print(f"Extracting features with {model_name}...")
        model, feature_dim = get_feature_extractor(model_name)
        features_list = []

        for images, _, filenames in tqdm(dataloader):
            images = images.to(DEVICE)
            outputs = model(images)

            # Flatten feature vector
            features = outputs.view(outputs.size(0), -1).cpu().numpy()[0]
            row = {"ImageName": filenames[0]}
            row.update({f"{model_name}_feature_{i+1}": feat for i, feat in enumerate(features)})
            features_list.append(row)

        df = pd.DataFrame(features_list)
        feature_dfs.append(df)

# Merge all features into one dataframe
from functools import reduce
final_df = reduce(lambda left, right: pd.merge(left, right, on="ImageName"), feature_dfs)
```

```python
final_df.insert(
    1,  # Position: second column
    "Label",
    final_df["ImageName"].apply(lambda name: name.split("_")[0])
)
print(final_df.head())
```

```python
# Save the final DataFrame to a CSV file
output_csv_path = os.path.join(DATASET_PATH, "image_features.csv")
final_df.to_csv(output_csv_path, index=False)
```
